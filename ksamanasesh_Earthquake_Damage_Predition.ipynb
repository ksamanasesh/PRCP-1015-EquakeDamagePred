# **Earthquake Damage Prediction**

## **Overall Report**

### **TASK 1 - Data Analysis Report**


**1. Data Overview :**

* The dataset contains 260,601 building records and 39 feature columns. After merging with the labels, the final dataset contains 40 columns.
* The dataset contains building, structural, geographic, and secondary-use information. The target variable is `damage_grade`, with three classes: `1`, `2`, and `3`.
* The dataset contains no missing values or duplicate records, and all 260,601 `building_id` values are unique.

**2. Categorical Features :**

* The main categorical features include `land_surface_condition`, `foundation_type`, `roof_type`, `ground_floor_type`, `other_floor_type`, `position`, `plan_configuration`, and `legal_ownership_status`.
* `foundation_type` showed a strong difference in damage distribution, with `foundation_type_i` having approximately 2.09% Grade 3 and `foundation_type_r` having approximately 37.85% Grade 3.
* Several categorical features contained rare categories, so their individual damage distributions were interpreted cautiously.

**3. Numerical Features :**

* The main numerical features analyzed were `count_floors_pre_eq`, `age`, `area_percentage`, and `height_percentage`.
* `count_floors_pre_eq` was mainly concentrated between 1 and 3 floors, while building age showed an increasing Grade 3 proportion across commonly represented age groups.
* `area_percentage` ranged from 1 to 100, with lower area groups showing a higher proportion of Grade 3 buildings.

**4. Binary Features :**

* The dataset contains binary features describing structural characteristics and secondary uses, including `has_superstructure_mud_mortar_stone`, `has_superstructure_timber`, `has_superstructure_bamboo`, and `has_secondary_use`.
* `has_superstructure_mud_mortar_stone` was the most common binary structural feature, appearing in 198,561 buildings.
* Several binary features showed noticeable differences in Grade 3 proportions between buildings where the feature was present and absent.

**5. Outliers :**

* Some numerical features contained extreme or very rare values. For example, `count_floors_pre_eq` contained 8 and 9 floors with only one observation each.
* The `age` feature contained an extreme value of 995, while some very high age values had very few observations.
* These rare values were not automatically removed because they may represent valid dataset records and removing them without evidence could alter the original data.

**6. Overall Analysis :**

* The analysis showed that foundation, ground-floor, roof, superstructure, building age, and geographic characteristics had noticeable relationships with `damage_grade`.
* The target variable was imbalanced, with Grade 2 representing 56.89% of the observations, followed by Grade 3 at 33.47% and Grade 1 at 9.64%.
* Overall, the analysis indicates that earthquake damage is associated with multiple building and geographic characteristics rather than a single feature.

### **TASK 2 - Predictive Modeling**


**1. Data Preprocessing :**

* The target variable `damage_grade` was separated from the input features. The `building_id` column and temporary EDA grouping columns were excluded from model training.
* The categorical features were converted into numerical form using one-hot encoding, resulting in **61 features** after encoding.
* The dataset was divided into training and testing sets using an 80:20 split with stratification to maintain the original `damage_grade` distribution.

**2. Model Building :**

* Three classification models were developed to predict `damage_grade`: `Decision Tree`, `Random Forest`, and `XGBoost`.
* The models were trained using the same training and testing datasets to provide a fair comparison of their predictive performance.
* The models were evaluated using accuracy, precision, recall, F1-score, and confusion matrices.

**3. Model Performance :**

* The `Decision Tree` model achieved an accuracy of **64.49%**, while the `Random Forest` model achieved **72.00%** accuracy.
* The `XGBoost` model achieved the highest accuracy of **72.21%**, slightly outperforming Random Forest.
* The results indicate that ensemble-based models performed better than the individual Decision Tree model for this dataset.

**4. Final Model Evaluation :**

* The selected `XGBoost` model achieved **72.21% accuracy**, **71.81% macro precision**, **62.44% macro recall**, and a **65.62% macro F1-score**.
* The model achieved a Mean Absolute Error of **0.2802**, indicating that the predicted damage grades were generally close to the actual grades.
* The confusion matrix showed that most classification errors occurred between the adjacent damage grades, particularly Grade 2 and Grade 3.

**5. Feature Importance :**

* The XGBoost feature-importance analysis identified `foundation_type_r` as the most important feature, with an importance score of **0.3373**.
* Other important features included `ground_floor_type_v`, `has_superstructure_mud_mortar_stone`, `foundation_type_i`, `geo_level_1_id`, and `has_superstructure_cement_mortar_brick`.
* These feature-importance values indicate predictive contribution within the model and should not be interpreted as direct causal relationships.

**6. Model Selection :**

* Among the three evaluated models, `XGBoost` achieved the highest test accuracy of **72.21%** and was selected as the final predictive model.
* `Random Forest` performed very closely with an accuracy of **72.00%**, while `Decision Tree` achieved a lower accuracy of **64.49%**.
* Therefore, `XGBoost` was selected for predicting earthquake building damage based on its overall predictive performance.

### **TASK 3 - Suggestion Report to Seismologists**

* **Improve foundation quality:**  Foundation-related features were among the most important predictors in the XGBoost model. Proper foundation design, construction, and structural assessment should therefore receive greater attention in earthquake-prone regions.

* **Strengthen ground-floor structures:** `ground_floor_type` showed noticeable differences in damage-grade distributions. Ground-floor construction should be carefully assessed and designed according to earthquake-resistant structural standards.

* **Improve structural materials:** Superstructure-related features showed meaningful relationships with `damage_grade`. The use of appropriate structural materials and earthquake-resistant construction techniques should be encouraged.

* **Prioritize vulnerable buildings:** Older and structurally vulnerable buildings should be identified through building inspections and prioritized for strengthening or retrofitting in areas with higher earthquake risk.

* **Consider geographic risk:** Geographic features such as `geo_level_1_id` were useful predictive features in the XGBoost model. Regional seismic risk should therefore be considered when planning construction standards and disaster-mitigation strategies.

* **Use machine learning for risk assessment:** Predictive models such as XGBoost can be used as supplementary tools to identify buildings or regions with potentially higher damage risk and help prioritize detailed engineering assessments.

* **Promote earthquake-resistant construction:** Building regulations and construction practices should emphasize earthquake-resistant design principles, particularly in regions exposed to significant seismic hazards.

### **MODEL COMPARISON REPORT**

**Model Comparison :**

* Three classification models, `Decision Tree`, `Random Forest`, and `XGBoost`, were trained and evaluated using the same training and testing datasets.

* The `Decision Tree` model achieved an accuracy of **64.49%**, while the `Random Forest` model achieved **72.00%** accuracy. This shows that the ensemble approach of Random Forest provided a significant improvement over a single Decision Tree.

### **CHALLENGES FACED REPORT**

**1. Class imbalance:**

*   The target variable `damage_grade` was imbalanced, with Grade 2 representing the majority of observations while Grade 1 had considerably fewer samples.

**2. Categorical variables:**
* Several important building characteristics were categorical and contained obfuscated category labels. These variables required encoding before model training.

**3. High-cardinality geographic features:**
* The dataset contains geographic-level identifiers such as `geo_level_1_id`, `geo_level_2_id`, and `geo_level_3_id` that represent locations rather than continuous numerical measurements.

**4. Rare categories:**
* Some categorical values occurred very infrequently, making their individual damage distributions less reliable.

**5. Nonlinear relationships:**
* The relatively poor performance of the Decision Tree and Logistic Regression compared with ensemble models indicated that the relationships between building characteristics and `damage_grade` are complex.

**6. Model interpretation:**
* Feature importance indicates which variables were useful for prediction, but it does not establish that those variables directly cause earthquake damage.

### 6. Data Preperation for Modeling

**Spliting X and Y**

```python
X = df.drop(columns=['damage_grade', 'area_group', 'height_group'])
y = df['damage_grade']
```

```python
categorical_cols = X.select_dtypes(include='object').columns.tolist()
print("Identifying Categorical Columns: \n")
categorical_cols
```

**OneHot CODE**

```python
X_encoded = pd.get_dummies(
    X,
    columns=categorical_cols,
    drop_first=True
)
```

```python
print("Original shape:", X.shape)
print("Encoded shape:", X_encoded.shape)
```

```python
X_encoded.info()
```

**Train Test and Split**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

### 7. Random Forest Classification and Evaluation

```python

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)
```

```python
y_pred = rf_model.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=[1, 2, 3],
    yticklabels=[1, 2, 3]
)

plt.xlabel('Predicted Damage Grade')
plt.ylabel('Actual Damage Grade')
plt.title('Random Forest - Confusion Matrix')
plt.show()
```
